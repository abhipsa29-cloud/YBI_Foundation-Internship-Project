# Python Loops

```python
#for loop
sum = 0

for i in range(1, 6):
    sum = sum + i
    print("Number:", i)

print("Total:", sum)
```

```python
#while loop
i = 1
sum = 0

while i <= 5:
    print("Number:", i)
    sum = sum + i
    i = i + 1

print("Total:", sum)
```

```python
#multiplication table
num = 5

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

```python
#reverse number
num = 12345

while num > 0:
    digit = num % 10
    print(digit)
    num = num // 10
```
