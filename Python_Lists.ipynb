# Python Lists

```python
#create and print list
fruits = ["Apple", "Mango", "Banana"]

print(fruits)
```

```python
#Access list elements
fruits = ["Apple", "Mango", "Banana"]

print(fruits[0])
print(fruits[1])
print(fruits[2])
```

```python
#adding elements
fruits = ["Apple", "Mango"]

fruits.append("Banana")

print(fruits)
```

```python
#remove an element
fruits = ["Apple", "Mango", "Banana"]

fruits.remove("Mango")

print(fruits)
```

```python
#change an element
fruits = ["Apple", "Mango", "Banana"]

fruits[1] = "Orange"

print(fruits)
```

```python
#sorting
numbers = [50, 20, 40, 10, 30]

numbers.sort()

print(numbers)
```

```python
#reversing
numbers = [10, 20, 30, 40, 50]

numbers.reverse()

print(numbers)
```
