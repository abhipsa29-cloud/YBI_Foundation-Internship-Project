# Python Conditional Statement

```python
# if-elif-else statement
marks = 75

if marks >= 80:
    print("A")
elif marks >= 60:
    print("B")
else:
    print("C")
```

```python
# Check positive & negative
num = 5

if num > 0:
    print("Positive")
else:
    print("Negative")
```
